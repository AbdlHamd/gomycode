```python
import pandas as pd
import numpy as np

# --- ÉTAPE 1 : Exploration des données avec Pandas ---
print("=== ÉTAPE 1 : Exploration des données avec Pandas ===\n")

# Chargement du dataset Diabetes (Pima Indians Diabetes Database)
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
columns = [
    "Pregnancies", "Glucose", "BloodPressure", "SkinThickness",
    "Insulin", "BMI", "DiabetesPedigreeFunction", "Age", "Outcome"
]
df = pd.read_csv(url, names=columns)

# 1. Vérification des premières lignes
print("1. Premières lignes du jeu de données :")
display(df.head())

# 2. Structure du jeu de données
print("\n2. Structure et types de données :")
df.info()

# 3. Identification des valeurs manquantes et des zéros aberrants
print("\n3. Analyse des valeurs manquantes et des zéros :(Dans ce dataset, les valeurs manquantes pour Glucose, BloodPressure, SkinThickness, Insulin et BMI sont souvent codées par 0)")
missing_stats = pd.DataFrame({
    'Valeurs_Nulles_NaN': df.isnull().sum(),
    'Valeurs_Egales_A_Zero': (df == 0).sum()
})
display(missing_stats)

# 4. Statistiques descriptives
print("\n4. Statistiques sommaires du jeu de données (describe) :")
display(df.describe())

# --- ÉTAPE 2 : Exploration avec ydata-profiling ---
print("\n=== ÉTAPE 2 : Exploration avec ydata-profiling ===")

# Installation silencieuse de ydata-profiling si nécessaire
!pip install -q ydata-profiling

from ydata_profiling import ProfileReport

# Génération du rapport interactif
profile = ProfileReport(df, title="Rapport d'Analyse du Diabète", explorative=True)

# Sauvegarde du rapport au format HTML dans l'environnement de travail
profile.to_file("rapport_diabete.html")
print("\n[INFO] Le rapport 'rapport_diabete.html' a été généré avec succès et est prêt à être téléchargé ou visualisé dans Colab.")

# --- DOCUMENTATION ET RECOMMANDATIONS ---
print("\n=== RÉSUMÉ DES OBSERVATIONS ET RECOMMANDATIONS ===\n")
print("""
Observations Clés :
- Zéros aberrants : Les colonnes 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', et 'BMI' contiennent des valeurs minimales égales à 0, ce qui est biologiquement impossible pour des patients vivants. Ces 0 correspondent à des valeurs manquantes masquées.
- Insulin & SkinThickness : Plus de 48% des valeurs pour 'Insulin' et environ 29% pour 'SkinThickness' sont à 0.
- Corrélations : La variable cible 'Outcome' (diabète) présente de fortes corrélations positives avec le niveau de 'Glucose', l'indice de masse corporelle ('BMI') et l'âge.

Prochaines étapes recommandées :
1. Imputation des valeurs manquantes : Remplacer les valeurs 0 aberrantes par la médiane ou la moyenne (par exemple, regroupée par groupe d'âge ou par Outcome) plutôt que de supprimer les lignes.
2. Traitement des valeurs aberrantes (Outliers) : Identifier et ajuster les valeurs extrêmement élevées de l'IMC ou de la pression artérielle.
3. Standardisation : Normaliser ou standardiser les variables explicatives avant d'entraîner un modèle prédictif (Machine Learning) car les échelles varient fortement (ex. DiabetesPedigreeFunction vs Insulin).
""")
```
